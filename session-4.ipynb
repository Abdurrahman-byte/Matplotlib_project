# Pie Chart, Gridlines & axis limits, Saving figures

```python
import matplotlib.pyplot as plt
# A pie chart is useful when you want to show a whole didvided into parts

subjects = ["Python", "Math", "Physics", "Other"]
hours = [10, 6, 8, 4]

plt.pie(hours, 
        labels=subjects,
        autopct="%1.1f%%",
        explode=[0, 0.1, 0, 0.1],
        shadow=True)

plt.title("Weekly Study Time")

plt.savefig("pie chart.png")

plt.show()
```

# gridlines

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([20, 40, 60, 80, 100])

plt.plot(x, 
         y,
         color="red",
         linewidth=2,
         marker="o",
         linestyle="--")

plt.xlabel("Hours")
plt.ylabel("Score")
plt.title("Study Progress")

plt.grid(True, linestyle="--")

plt.show()
```

# Axis Limit

```python
# Axis limit controls what part of the graph is visible
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([20, 40, 60, 80, 100])

plt.plot(x,
         y, 
         linestyle="--",
         linewidth=2,
         marker="o")

plt.xlim(0, 10)
plt.ylim(0, 100)

plt.xlabel("hours")
plt.ylabel("Score")
plt.title("Study progress")
plt.grid(True,
         linestyle="--")

plt.show()
```
