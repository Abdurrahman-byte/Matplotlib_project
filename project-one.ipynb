```python
import matplotlib.pyplot as plt 
import pandas as pd
```

# Project 1

```python
data = {
    "Student": ["Ali", "Fatima", "John", "Sarah", "David"],
    "Hours_Studied": [2, 4, 6, 8, 10],
    "Attendance": [70, 82, 90, 95, 98],
    "Score": [55, 68, 78, 88, 94]
}

df = pd.DataFrame(data)
print(df.head())
print(df.info())

plt.bar(df["Student"], df["Score"])
plt.xlabel("Student")
plt.ylabel("Score")
plt.title("Student Exams Score")

plt.show()
```

```python
plt.scatter(df["Hours_Studied"], 
            df["Score"],
            marker="o",
            linestyle="--")
plt.xlabel("Hour Studied")
plt.ylabel("Score")
plt.title("Hour Studied vs  Exams Score")

plt.show()
```

```python
plt.hist(df["Score"], bins=5)
plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Exams Scores")

plt.show()
```
