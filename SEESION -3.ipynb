```python
import matplotlib.pyplot as plt
```

# SCATTER PLOTS, BAR CHART, SUBPLOTS

```python
hours = [1, 2, 3, 4, 5, 6]
scores = [50, 55, 62, 70, 78, 88]

plt.scatter(hours, scores)

plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.title("Study Hours Vs Exam Score")

plt.show()
```

Scatter plots are use for seeing the relationship between two numerical variables

plots() vs scatter()
plt.plot() - shows the trnd/path
plt.scatter() - show me the individaul data points

```python
languages = ["python", "Javascript", "C++", "java"]
students = [80, 65, 40, 55]

plt.bar(languages, students)

plt.xlabel("programming language")
plt.ylabel("Number of students")
plt.title("Programming language popularity")

plt.show()
```

Bar chart are for comparing categories

```python

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.subplot(1, 2, 1)

plt.plot(x, y)
plt.title("Line Plot")


plt.subplot(1, 2, 2)

plt.scatter(x, y)
plt.title("Scatter Plot")

plt.show()
```

Subplot() means diving your figure into a graph slots
subplot mental model: subplot(rows, column, position)

```python
hours = [2, 4, 6, 8]
scores = [50, 75, 45, 90]

plt.scatter(hours, scores)

plt.xlabel('Study session')
plt.ylabel('Exam score')
plt.title('Study Hour Vs Exam Score')

plt.show()
```

```python
languages = ["Python", "javascript", "C++", "Java"]
users = [90, 65, 40, 24]

plt.bar(languages, users)

plt.xlabel("Programming language")
plt.ylabel("Students")
plt.title("Programming language popularity")

plt.show()
```

```python
a = [1, 2, 3, 4, 5]
b = [2, 4, 6, 8, 10]

plt.plot(a, b)
plt.grid()

plt.xlabel("Coordinate A")
plt.ylabel("Coordinate B")
plt.title("A Vs B")

plt.show()
```

```python
planet_orbital_period = [87, 224, 365, 686, 4332, 10759, 30688, 60182]
orbital_velocity = [47, 35, 29, 24, 13, 9, 6,  5]

plt.scatter(planet_orbital_period, orbital_velocity)
plt.grid()

plt.xlabel("Earth days")
plt.ylabel("Orbital velocity")
plt.title("Planets Orbit around the Sun")

plt.show()
```
