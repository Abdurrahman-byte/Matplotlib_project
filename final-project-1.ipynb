```python
import matplotlib.pyplot as plt
import pandas as pd
```

# Using a CSV file for Analysis & Visualization

```python
df = pd.read_csv("ecommerce_sales_real_world_practice.csv")
print(df.info())
print(df.head())
print(df.groupby("Product")["Units_Sold"].sum())
print(df.groupby("Region")["Revenue_USD"].sum())
print(df["Customer_Rating"].mean())
print(df.groupby("Product")["Returned_Units"].sum())
```

```python
sales_by_product = df.groupby("Product")["Units_Sold"].sum()
print(sales_by_product)
```

```python
plt.bar(sales_by_product.index,
        sales_by_product.values,
         edgecolor="black")
plt.xlabel("Product")
plt.ylabel("Total Units Sold")
plt.title("Sales History")

plt.show()
```

```python
revenue_by_region = df.groupby("Region")["Revenue_USD"].sum()
print(revenue_by_region)
```

```python
plt.bar(revenue_by_region.index,
        revenue_by_region.values)
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.title("Revenue Made Per Region")
plt.tight_layout()

plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
daily_revenue = df.groupby("Date")["Revenue_USD"].sum()

print(df["Date"])
print(daily_revenue)
```

```python
plt.plot(daily_revenue.index,
         daily_revenue.values)
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.title("Revenue Overtime")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
plt.scatter(df["Customer_Rating"], df["Units_Sold"])
plt.xlabel("Customer")
plt.ylabel("Unit Sold")
plt.title("Relationship")
plt.tight_layout()

plt.show()
```
