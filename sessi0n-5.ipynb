```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.plot([1, 2, 3, 4], [2, 4, 6, 8])

ax.set_xlabel("X label")
ax.set_ylabel("Y label")
ax.set_title("my graph")

plt.show()
```

 figure is the entire canvas/window containing your visualization

 Axes is the actual plotting area 

# Multiple Lines & styling

```python
import matplotlib.pyplot as plt

programming_languages = [1, 2, 3, 4, 5]

python = [50, 89, 49, 75, 69]
Javascript = [40, 50, 70, 65, 64]

plt.plot(programming_languages,
         python,
         label="Python",
         marker="x",
         linestyle="--")

plt.plot(programming_languages,
         Javascript,
         label="Javascript",
         marker="o",
         linestyle="--")


plt.legend
plt.show()
```

# Annotations

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [39, 49, 59, 65]

plt.plot(x, y)

plt.annotate(
    "Highest score",
    xy=(4, 65),
    xytext=(3.5, 75),
    arrowprops=dict(arrowstyle="->")
)

plt.xlabel("Students")
plt.ylabel("Students Score")
plt.title("Exams Performance")

plt.show()
```

# Histograms

Histograms are to show the distribution of numerical data

Bar chart vs Histogram

bar chart => compares categories

histogram => shows the distribution of numerical data

bins determine how many groups/ranges the data gets didvided into

```python
import matplotlib.pyplot as plt

scores = [
    45, 52, 56, 61, 62, 65, 67, 68, 70,
    71, 72, 73, 74, 75, 76, 78, 80, 82,
    85, 90
]

plt.hist(scores, bins=5)

plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Exams Score")

plt.show()
```

# Making Figures Cleaner

controls the width and height of the figure

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot([1, 2, 3, 4], [10, 20, 30, 40])

plt.title("Progress")
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 2, figsize=(10, 7))

plt.tight_layout()
plt.show()
```

figsize control the size of the canvas
tight_layout() organizes the spacing inside the canvas
