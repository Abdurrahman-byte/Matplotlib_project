```python
import pandas as pd
import matplotlib.pyplot as plt
```

# Inspecting Dataset

```python
df = pd.read_csv("ecommerce_sales_real_world_practice.csv")
df["Date"] = pd.to_datetime(df["Date"])
```

# Aggregations

```python
total_revenue = df.groupby("Product")["Revenue_USD"].sum()
sortd = total_revenue.sort_values(ascending=False)
regional_breakdown = df["Customer_Rating"].mean()
total_unit_sold = df.groupby("Region")["Units_Sold"].sum()
df["Returned_Rate%"] = (df["Returned_Units"] / df["Units_Sold"] * 100)

print(sortd)
print(regional_breakdown)
print(total_unit_sold)
print(df["Returned_Rate%"])
```

# Category Revenue

```python
category_revenue = df.groupby("Category")["Revenue_USD"].sum()
```

# Category Revenue Bar Chart

```python
plt.bar(category_revenue.index,
        category_revenue.values,
        color="Blue")

plt.xlabel("Product")
plt.ylabel("Revune")
plt.grid(axis="y")
plt.title("Total Revenue")

plt.show()
```

# Revenue Overtime & Line Plot

```python
revenue_overtime = df.groupby("Date")["Revenue_USD"].sum()
print(revenue_overtime)
```

```python
plt.plot(revenue_overtime.index,
         revenue_overtime.values,
         marker="o")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
