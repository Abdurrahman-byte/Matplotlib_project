```python
import matplotlib.pyplot as plt
```

# Labels Titles & Legends

```python
hours = [1, 2, 3, 4, 5]
scores = [50, 55, 65, 72, 85]

plt.plot(hours, scores)

plt.xlabel("Study Hour")
plt.ylabel("Exam Score")
plt.title("study Hours vs Exams score")

plt.show()
```

# Adding legend

```python
month = [1, 2, 3, 4, 5]

python = [60, 70, 75, 85, 90]
Javascript = [80, 82, 85, 87, 89]

plt.plot(month, python, label="python")
plt.plot(month, Javascript, label="Javascript")

plt.xlabel("Month")
plt.ylabel("Skill Score")
plt.title("Programming progress")

plt.legend()

plt.show()
```

# Basic customization

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(
    x,
    y,
    color="red",
    #changes the color
    linestyle="--",
    #makes the line dashed
    marker="o",
    #puts a circle marker on each data point
    linewidth=2
    #controls how thick the line is 
)

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("My first customized Graph")

plt.show()
```
