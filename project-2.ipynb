```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Product": ["Laptop", "Phone", "Headphones", "Monitor", "Keyboard"],
    "Sales": [15, 32, 45, 20, 38],
    "Revenue": [15000, 22400, 45000, 6000, 3800]
}

df = pd.DataFrame(data)
print(df.info())
print(df.head())
print(df["Sales"].max())
```

# Project 2

finding the highest number of sales

```python
plt.bar(df["Product"], df["Sales"])
plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product Sales")

plt.show()
```

```python
plt.scatter(df["Sales"], df["Revenue"])
plt.xlabel("Sales")
plt.ylabel("Revenue")
plt.title("Sales vs Revenue")

plt.show()
```
