```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Python": [60, 70, 75, 85, 90],
    "Javascript": [80, 82, 85, 87, 89]
}

df = pd.DataFrame(data)
```

```python
plt.plot(df["Month"], df["Python"], label="Python", marker="x")
plt.plot(df["Month"], df["Javascript"], label="Javscript", marker="o")
plt.xlabel("Month")
plt.ylabel("Score")
plt.title("python progress")

plt.legend()
plt.show()
```
